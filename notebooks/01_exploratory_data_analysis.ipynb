# Target Variable Analysis

In this section, we analyze the distribution of attack categories in the training dataset.

Understanding class imbalance is essential before selecting evaluation metrics and machine learning algorithms.

```python
attack_counts = train_df["label"].value_counts()

attack_counts
```

```python
plt.figure(figsize=(14,6))

attack_counts.plot(kind="bar")

plt.title("Attack Distribution")
plt.xlabel("Attack Type")
plt.ylabel("Number of Samples")

plt.xticks(rotation=60)

plt.show()
```

```python
px.bar(
    attack_counts,
    title="Attack Distribution",
    labels={
        "value": "Count",
        "index": "Attack Type"
    }
)
```

### Observation

- The dataset is highly imbalanced.
- The majority of samples belong to the **normal** and **neptune** classes.
- Several attack categories contain very few samples.
- Accuracy alone will not be an appropriate evaluation metric.

# Protocol Analysis

```python
protocol_counts = train_df["protocol_type"].value_counts()

protocol_counts
```

```python
plt.figure(figsize=(6,5))

protocol_counts.plot(
    kind="bar",
)

plt.title("Protocol Distribution")

plt.show()
```

```python
px.pie(
    values=protocol_counts.values,
    names=protocol_counts.index,
    title="Protocol Distribution"
)
```

### Observation

- TCP dominates the dataset.
- UDP appears significantly less frequently.
- ICMP traffic represents the smallest portion of the dataset.

# Missing Values Analysis

Before training any machine learning model, we verify whether the dataset contains missing values.

Missing values can negatively affect model performance and often require preprocessing strategies such as imputation or removal.

```python
missing_values = train_df.isnull().sum()

missing_values[missing_values > 0]
```

### Observation

No missing values were found in the training dataset.

No imputation or missing-value handling will be required.

# Correlation Analysis

Correlation helps identify relationships between numerical features.

Highly correlated variables may contain redundant information.

```python
correlation_matrix = train_df.corr(numeric_only=True)

correlation_matrix
```

```python
correlations = (
    correlation_matrix
    .abs()
    .unstack()
    .sort_values(ascending=False)
)

correlations = correlations[
    correlations < 1
]

correlations.head(20)
```
